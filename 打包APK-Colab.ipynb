## 第 2 步：上传项目压缩包
把 `nfc-app` 文件夹压缩成 zip（至少包含 main.py 和 buildozer.spec），选择 zip 文件上传。

```python
from google.colab import files
import zipfile, os

uploaded = files.upload()          # 弹出选择框, 选 nfc-app.zip
zip_name = list(uploaded.keys())[0]
with zipfile.ZipFile(zip_name) as z:/n    z.extractall('project')

# 兼容两种压缩方式(带/不带顶层文件夹)
if os.path.exists('project/main.py'):
    os.chdir('project')
elif len(os.listdir('project')) == 1:
    os.chdir(os.path.join('project', os.listdir('project')[0]))
print('当前目录:', os.getcwd())
print('项目文件:', os.listdir('.'))
assert 'main.py' in os.listdir('.') and 'buildozer.spec' in os.listdir('.'), '项目不完整!'
```

## 第 3 步：开始打包（首次约 25-40 分钟，会自动下载 Android SDK/NDK，请耐心等待）

```python
!yes | buildozer android debug 2>&1 | tail -30
import glob
apks = glob.glob('bin/*.apk')
print('\n✅ 打包完成:', apks if apks else '❌ 未找到 APK, 请检查上方报错')
```

## 第 4 步：下载 APK 到电脑

```python
from google.colab import files
import glob
apk = glob.glob('bin/*.apk')[0]
files.download(apk)   # 浏览器开始下载 APK
```

## 第 5 步：安装到手机
1. 把 APK 传到手机（微信文件传输助手 / 数据线均可）
2. 手机上点击 APK 安装（需允许"安装未知应用"）
3. 打开"NFC 助手"，确保手机 NFC 已开启，即可读写 NFC 标签
